```python
# ---------- 0. DEPS (safe to re-run; fresh Colab runtimes have none of these) ----------
import importlib, subprocess, sys

def ensure(mod, pkg=None):
    try:
        importlib.import_module(mod)
        return False
    except ImportError:
        print('installing', pkg or mod, '...')
        subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', pkg or mod], check=True)
        return True

for _m, _p in [('crewai', 'crewai'), ('litellm', 'litellm'), ('nest_asyncio', 'nest_asyncio')]:
    ensure(_m, _p)
importlib.invalidate_caches()
print('deps ready')

# ==============================================================
# CONTENT ENGINE v1 - clean rebuild
# Trend Scout -> Creative Strategist -> Content Creator & SEO
# ==============================================================
import os, json, datetime, textwrap
from typing import List

import nest_asyncio, litellm
nest_asyncio.apply()
litellm.drop_params = True          # Groq rejects cache_breakpoint / cache_control

# CrewAI 1.15 injects a 'cache_breakpoint' key INSIDE each message dict. Groq rejects it,
# and litellm.drop_params only filters top-level params - which is why that error never went away.
# Strip it at the litellm boundary so every crewai call is clean.
def _strip_cache_keys(obj):
    if isinstance(obj, dict):
        return {k: _strip_cache_keys(x) for k, x in obj.items()
                if k not in ('cache_breakpoint', 'cache_control')}
    if isinstance(obj, list):
        return [_strip_cache_keys(x) for x in obj]
    return obj

if not getattr(litellm, '_cb_patched', False):
    _orig_completion = litellm.completion
    _orig_acompletion = litellm.acompletion

    def _patched_completion(*args, **kwargs):
        if 'messages' in kwargs:
            kwargs['messages'] = _strip_cache_keys(kwargs['messages'])
        return _orig_completion(*args, **kwargs)

    async def _patched_acompletion(*args, **kwargs):
        if 'messages' in kwargs:
            kwargs['messages'] = _strip_cache_keys(kwargs['messages'])
        return await _orig_acompletion(*args, **kwargs)

    litellm.completion = _patched_completion
    litellm.acompletion = _patched_acompletion
    try:
        import litellm.main as _lmain
        _lmain.completion = _patched_completion
        _lmain.acompletion = _patched_acompletion
    except Exception:
        pass
    litellm._cb_patched = True
    print('litellm patched: cache_breakpoint will be stripped from messages')

from pydantic import BaseModel, Field
from crewai import Agent, Task, Crew, Process, LLM

# ---------- 1. KEYS (Colab Secrets only - never hardcode) ----------
from google.colab import userdata

def secret(name):
    try:
        v = userdata.get(name)
        return v.strip() if v else None
    except Exception:
        return None

GROQ_KEY = secret("GROQ_API_KEY")
assert GROQ_KEY and GROQ_KEY.startswith("gsk_"), (
    "GROQ_API_KEY missing or malformed in Colab Secrets. "
    "Open the key icon in the left sidebar, add GROQ_API_KEY, enable notebook access."
)
os.environ["GROQ_API_KEY"] = GROQ_KEY

SERPER_KEY = secret("SERPER_API_KEY")
if SERPER_KEY:
    os.environ["SERPER_API_KEY"] = SERPER_KEY
USE_SERPER = bool(SERPER_KEY)

# ---------- 2. BRAND CONFIG - swap this block per client ----------
BRAND = {
    "name": "Alphachain Crypto",
    "industry": "crypto and web3 education",
    "audience": "Indian retail crypto investors, 22-35, mobile-first",
    "voice": "sharp, contrarian, plain English, zero hype",
    "platform": "Instagram",
    "cta": "Follow @Alpha_chain_ for the alpha",
    "avoid": "financial advice, guaranteed returns, emoji in the hook",
}
NICHE = BRAND["industry"]

# ---------- 3. LLM ----------
llm = LLM(model="groq/openai/gpt-oss-120b", temperature=0.7, max_tokens=2000)

# fail fast with a readable message instead of a crew traceback
_probe = litellm.completion(
    model="groq/openai/gpt-oss-120b",
    messages=[{"role": "user", "content": "ping"}],
    max_tokens=5,
)
print("Groq key OK ->", _probe.choices[0].message.content[:20])

# ---------- 4. TOOLS ----------
tools = []
if USE_SERPER:
    try:
        try:
            from crewai_tools import SerperDevTool
        except ImportError:
            import subprocess, sys
            print("installing crewai-tools ...")
            subprocess.run([sys.executable, "-m", "pip", "install", "-q", "crewai-tools"], check=True)
            from crewai_tools import SerperDevTool
        tools = [SerperDevTool(n_results=10)]
        print("Serper web search: ON")
    except Exception as e:
        print("Serper tool unavailable, falling back to model knowledge:", e)
else:
    print("Serper web search: OFF (no SERPER_API_KEY) - scout runs on model knowledge")

# ---------- 5. STRUCTURED OUTPUT ----------
class PostPackage(BaseModel):
    hook: str = Field(description="First line of the caption. Under 12 words. No emoji.")
    caption: str = Field(description="Full Instagram caption, 80-150 words, line breaks allowed.")
    hashtags: List[str] = Field(description="Exactly 12 hashtags, no # duplicates.")
    visual_direction: str = Field(description="Art direction brief for the Canva creative.")
    alt_text: str = Field(description="Accessibility alt text under 125 characters.")
    trend_hook_source: str = Field(description="Which trend this post is riding, one line.")
    seo_keywords: List[str] = Field(description="5 search keywords this post targets.")

# ---------- 6. AGENTS (memory=False: memory=True needs an embedding key) ----------
trend_scout = Agent(
    role="Trend Scout",
    goal=f"Surface what is genuinely trending right now in {NICHE} that {BRAND['audience']} actually cares about.",
    backstory=(
        "You are a Trend Scout with a decade of pattern recognition across social platforms. "
        "You reject evergreen filler and generic advice. You only report trends with a specific, "
        "datable hook, and you say plainly when a trend is fading."
    ),
    tools=tools, llm=llm, verbose=True, memory=False, allow_delegation=False, max_iter=6,
)

creative_strategist = Agent(
    role="Creative Strategist",
    goal=f"Turn raw trends into one sharp {BRAND['platform']} content angle that fits {BRAND['name']}'s voice.",
    backstory=(
        "You are a strategist who has shipped thousands of performing posts. You pick ONE angle and "
        "defend it. You know the difference between a topic and an angle, and you kill ideas that "
        "any competitor could have posted."
    ),
    llm=llm, verbose=True, memory=False, allow_delegation=False, max_iter=6,
)

content_creator = Agent(
    role="Content Creator and SEO Optimizer",
    goal="Write the finished post package: hook, caption, hashtags, visual direction, alt text, keywords.",
    backstory=(
        "You write captions people finish reading. You front-load the hook, keep sentences short, and "
        "treat hashtags as search surface rather than decoration. You never pad."
    ),
    llm=llm, verbose=True, memory=False, allow_delegation=False, max_iter=6,
)

# ---------- 7. TASKS ----------
scout_task = Task(
    description=textwrap.dedent(f"""
        Identify the 5 strongest current trends, conversations or viral formats in {NICHE}
        that would land with: {BRAND['audience']}.
        For each one give: the trend, why it is hot right now, the emotional driver,
        and how a brand could enter the conversation without looking late.
        Be specific. No generic advice like 'educational content performs well'.
    """).strip(),
    expected_output="A numbered list of 5 trends, each with trend / why now / emotional driver / brand entry point.",
    agent=trend_scout,
)

strategy_task = Task(
    description=textwrap.dedent(f"""
        From the scouted trends, pick the SINGLE best angle for {BRAND['name']}.
        Brand voice: {BRAND['voice']}. Platform: {BRAND['platform']}. Avoid: {BRAND['avoid']}.
        Output: the chosen trend, the specific angle, why this beats the other four,
        the one feeling the post should produce, and a one-line content brief for the writer.
    """).strip(),
    expected_output="A single chosen angle with justification and a one-line brief.",
    agent=creative_strategist,
    context=[scout_task],
)

creation_task = Task(
    description=textwrap.dedent(f"""
        Write the complete {BRAND['platform']} post package from the strategist's brief.
        Brand: {BRAND['name']}. Voice: {BRAND['voice']}. CTA to close with: {BRAND['cta']}.
        Hard rules: hook under 12 words and no emoji in it; caption 80-150 words;
        exactly 12 hashtags mixing broad, niche and branded; alt text under 125 characters.
        Avoid: {BRAND['avoid']}.
    """).strip(),
    expected_output="A PostPackage object with every field populated.",
    agent=content_creator,
    context=[strategy_task],
    output_pydantic=PostPackage,
)

# ---------- 8. RUN ----------
crew = Crew(
    agents=[trend_scout, creative_strategist, content_creator],
    tasks=[scout_task, strategy_task, creation_task],
    process=Process.sequential,
    verbose=True,
    memory=False,
)

import asyncio
try:
    asyncio.get_running_loop()
    _in_loop = True
except RuntimeError:
    _in_loop = False

started = datetime.datetime.utcnow()
if _in_loop:
    # Colab cells already run inside an event loop; CrewAI 1.15 refuses sync kickoff there
    result = await crew.kickoff_async()
else:
    result = crew.kickoff()
finished = datetime.datetime.utcnow()

# ---------- 9. PERSIST (this JSON is what the dashboard reads) ----------
post = result.pydantic.model_dump() if result.pydantic else {"raw": str(result)}

run = {
    "run_id": f"run_{started.strftime('%Y%m%d_%H%M%S')}",
    "brand": BRAND["name"],
    "platform": BRAND["platform"],
    "started_at": started.isoformat() + "Z",
    "finished_at": finished.isoformat() + "Z",
    "duration_s": round((finished - started).total_seconds(), 1),
    "status": "success",
    "serper_enabled": USE_SERPER,
    "model": "groq/openai/gpt-oss-120b",
    "stages": [
        {"agent": "Trend Scout", "status": "success", "output": str(scout_task.output)},
        {"agent": "Creative Strategist", "status": "success", "output": str(strategy_task.output)},
        {"agent": "Content Creator and SEO Optimizer", "status": "success", "output": str(creation_task.output)},
    ],
    "post": post,
    "usage": getattr(result, "token_usage", None).__dict__ if getattr(result, "token_usage", None) else None,
}

with open("/content/run_latest.json", "w") as f:
    json.dump(run, f, indent=2, default=str)

print("\n" + "=" * 60)
print("RUN COMPLETE ->", run["run_id"], f"({run['duration_s']}s)")
print("=" * 60)
print(json.dumps(post, indent=2))

# ---------- 10. OPTIONAL: push to the Vercel dashboard ----------
DASHBOARD_URL = secret("DASHBOARD_URL")   # e.g. https://your-app.vercel.app/api/runs
DASHBOARD_TOKEN = secret("DASHBOARD_TOKEN")
if DASHBOARD_URL:
    import requests
    r = requests.post(
        DASHBOARD_URL, json=run,
        headers={"Authorization": f"Bearer {DASHBOARD_TOKEN}"} if DASHBOARD_TOKEN else {},
        timeout=20,
    )
    print("Dashboard POST ->", r.status_code, r.text[:200])
else:
    print("DASHBOARD_URL not set - skipped push. Add it to Colab Secrets after deploying.")
```
